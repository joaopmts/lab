```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
    .appName("check spotify tracks output")\
    .master("spark://spark-master:7077")\
    .getOrCreate()

spark
```

```python
df = spark.read.parquet("s3a://temp/spotify")
```

```python

# Mostra algumas linhas
df.show(5, truncate=False)

# Conta registros
print("Total de linhas:", df.count())

# Verifica duplicados por id
dupes = df.groupBy("id").count().filter("count > 1")
dupes.show()
```

```python
df.count() - df.dropDuplicates(["id"]).count()
```
