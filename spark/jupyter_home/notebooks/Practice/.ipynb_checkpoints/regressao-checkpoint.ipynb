```python
from pyspark.sql import SparkSession, functions as f
from pyspark.sql.types import IntegerType, DoubleType
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pyspark.ml.regression import LinearRegression
```

```python
spark = SparkSession.builder\
    .master("spark://spark-master:7077")\
    .appName("Regressao com Spark")\
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

spark
```

```python
imoveis = spark.read.json("s3a://bronze/regressao")
```

```python
#seleção colunas
dataset = imoveis\
    .select('ident.customerID', 'listing.types.*', 'listing.features.*', 'listing.address.*', 'listing.prices.price', 'listing.prices.tax.*')\
    .drop('city', 'location', 'totalAreas')

#alterar tipo
dataset = dataset\
    .withColumn('usableAreas', dataset['usableAreas'].cast(IntegerType()))\
    .withColumn('price', dataset['price'].cast(IntegerType()))\
    .withColumn('condo', dataset['condo'].cast(IntegerType()))\
    .withColumn('iptu', dataset['iptu'].cast(IntegerType()))\

dataset.printSchema()

dataset.show()
```

```python
dataset\
    .select('usage')\
    .groupBy('usage')\
    .agg(f.count('usage'))\
    .show()

dataset\
    .select('zone')\
    .groupBy('zone')\
    .agg(f.count('zone'))\
    .show()

dataset.select([
    f.count(
        f.when(f.col(c).isNull() | f.isnan(c), c)
    ).alias(c)
    for c in dataset.columns
]).show()
```

```python
#remoção de tipo comercial
dataset = dataset\
    .select('*')\
    .filter(dataset['usage'] == 'Residencial')

#fill 0 para campos na
dataset = dataset\
    .select('*')\
    .na\
    .fill(0)

#remoção zona vazia
dataset = dataset\
    .select('*')\
    .filter(dataset['zone'] != '')
```

```python
#variavel dummy
unit = dataset\
    .groupBy('customerID')\
    .pivot('unit')\
    .agg(f.lit(1))\
    .na.fill(0)

zone = dataset\
    .groupBy('customerID')\
    .pivot('zone')\
    .agg(f.lit(1))\
    .na.fill(0)

dataset = dataset\
    .join(unit, 'customerID', how='inner')\
    .join(zone, 'customerID', how='inner')

```

```python
#setar variavel para ml
dataset = dataset.withColumnRenamed('price', 'label')

X = [
    'bathrooms',
    'bedrooms',
    'floors',
    'parkingSpaces',
    'suites',
    'unitFloor',
    'unitsOnTheFloor',
    'usableAreas',
    'condo',
    'iptu',
    'Apartamento',
    'Casa',
    'Outros',
    'Zona Central',
    'Zona Norte',
    'Zona Oeste',
    'Zona Sul'
]
#montando vetor para ml
assembler = VectorAssembler(inputCols = X, outputCol = 'features')
dataset_ml = assembler.transform(dataset).select('features', 'label')
```

```python
#fazendo matriz de correlacao e selectionando apenas pearson
correlation_df = Correlation.corr(dataset_ml, 'features').collect()[0][0]
#ou
#correlation_df = Correlation.corr(dataset_ml, 'features', method='pearson').collect()[0][0]

#tranformando para dataframe pandas
df_corr_array = pd.DataFrame(correlation_df.toArray(), columns=X, index=X)

#criando heatmap para melhor vizualizacao
plt.figure(figsize=(12,10))
paleta = sns.color_palette("light:salmon", as_cmap=True)
sns.heatmap(df_corr_array.round(1), annot=True, cmap=paleta)
plt.show()

#criando modelo ml
    #setando proporção e seed fixa
treino, teste = dataset_ml.randomSplit([0.7, 0.3], seed=101)
print("Treino:", treino.count())
print("Teste:", teste.count())

#criação do modelo
lr = LinearRegression()
modelo_lr = lr.fit(treino)

#ver aderencia
previsoes_lr_treino  = modelo_lr.transform(treino)
previsoes_lr_treino.show()
```

```python
resumo_treino = modelo_lr.summary
print("R2:", resumo_treino.r2)
print("RMSE:", resumo_treino.rootMeanSquaredError)

resumo_teste = modelo_lr.evaluate(teste)
print("R2:", resumo_teste.r2)
print("RMSE:", resumo_teste.rootMeanSquaredError)
```
