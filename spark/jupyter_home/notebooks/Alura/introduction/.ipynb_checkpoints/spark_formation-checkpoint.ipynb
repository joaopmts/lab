# Lendo da S3

```python
company_df = spark.read.csv(path='s3a://bronze/receitafederal/empresas', sep=";", inferSchema=True)
store_df = spark.read.csv(path='s3a://bronze/receitafederal/estabelecimentos', sep=";", inferSchema=True)
bussiness_parter_df = spark.read.csv(path='s3a://bronze/receitafederal/socios', sep=";", inferSchema=True)
```

# Tratamentos

```python
# Schema colunas
estabsColNames = ['cnpj_basico', 'cnpj_ordem', 'cnpj_dv', 'identificador_matriz_filial', 'nome_fantasia', 'situacao_cadastral', 'data_situacao_cadastral', 'motivo_situacao_cadastral', 'nome_da_cidade_no_exterior', 'pais', 'data_de_inicio_atividade', 'cnae_fiscal_principal', 'cnae_fiscal_secundaria', 'tipo_de_logradouro', 'logradouro', 'numero', 'complemento', 'bairro', 'cep', 'uf', 'municipio', 'ddd_1', 'telefone_1', 'ddd_2', 'telefone_2', 'ddd_do_fax', 'fax', 'correio_eletronico', 'situacao_especial', 'data_da_situacao_especial']
empresasColNames = ['cnpj_basico', 'razao_social_nome_empresarial', 'natureza_juridica', 'qualificacao_do_responsavel', 'capital_social_da_empresa', 'porte_da_empresa', 'ente_federativo_responsavel']
sociosColNames = ['cnpj_basico', 'identificador_de_socio', 'nome_do_socio_ou_razao_social', 'cnpj_ou_cpf_do_socio', 'qualificacao_do_socio', 'data_de_entrada_sociedade', 'pais', 'representante_legal', 'nome_do_representante', 'qualificacao_do_representante_legal', 'faixa_etaria']
```

```python
#Renomeando colunas
for index, colname in enumerate(empresasColNames):
    company_df = company_df.withColumnRenamed(f"_c{index}", colname)

for index, colname in enumerate(estabsColNames):
    store_df = store_df.withColumnRenamed(f"_c{index}", colname)

for index, colname in enumerate(sociosColNames):
    bussiness_parter_df = bussiness_parter_df.withColumnRenamed(f"_c{index}", colname)
```

```python
company_df.printSchema()
store_df.printSchema()
bussiness_parter_df.printSchema()
```

```python
#Alterando tipos colunas

company_df = company_df\
    .withColumn(
        'capital_social_da_empresa',
        f.regexp_replace('capital_social_da_empresa', ',', '.')
        .cast(DoubleType())
)

store_df = store_df\
    .withColumn(
        'data_situacao_cadastral',
        f.to_date(store_df.data_situacao_cadastral.cast(StringType()), 'yyyyMMdd')
    )\
    .withColumn(
        'data_de_inicio_atividade',
        f.to_date(store_df.data_de_inicio_atividade.cast(StringType()), 'yyyyMMdd')
    )\
    .withColumn(
        'data_da_situacao_especial',
        f.to_date(store_df.data_da_situacao_especial.cast(StringType()), 'yyyyMMdd')
)\

bussiness_parter_df = bussiness_parter_df\
    .withColumn(
        'data_de_entrada_sociedade',
        f.to_date(bussiness_parter_df.data_de_entrada_sociedade.cast(StringType()), 'yyyyMMdd')
)
```

```python
company_df.printSchema()
store_df.printSchema()
bussiness_parter_df.printSchema()
```

```python
company_df\
    .select('cnpj_basico', 'razao_social_nome_empresarial', 'capital_social_da_empresa')\
    .show(5)
```

```python
bussiness_parter_df.printSchema()
```

```python
bussiness_parter_df\
    .select('nome_do_socio_ou_razao_social','faixa_etaria',
    f.year('data_de_entrada_sociedade').alias('ano_de_entrada'))\
    .show(5)
```

```python
#Replace
bussiness_parter_df\
    .select([f.count(f.when(f.isnull(i), 1)).alias(i)\
    for i in bussiness_parter_df.columns])\
    .show()
```

```python
#bussiness_parter_df.na.fill(0) - Substitui todo o valor NA de todo o dataframe por 0, como foi 0 apenas para numeros
#bussiness_parter_df.na.fill("-") - Substitui todo o valor NA de todo o dataframe por -, como foi - apenas para strings
#NaN <> Null - NaN - Not a Number, Null - Vazio 
```

```python
#Ordenação
bussiness_parter_df\
    .select('nome_do_socio_ou_razao_social','faixa_etaria',
    f.year('data_de_entrada_sociedade').alias('ano_de_entrada'))\
    .orderBy(['ano_de_entrada', 'faixa_etaria'], ascending=[False,False])\
    .show(5)
```

```python
#filtro
company_df\
    .where('capital_social_da_empresa==50')\
    .show(5, False)

bussiness_parter_df\
    .select('nome_do_socio_ou_razao_social')\
    .filter(bussiness_parter_df.nome_do_socio_ou_razao_social.startswith("JOAO"))\
    .filter(bussiness_parter_df.nome_do_socio_ou_razao_social.endswith("RODRIGUES"))\
    .show(5,False)

company_df\
    .select('razao_social_nome_empresarial', 'natureza_juridica', 'porte_da_empresa', 'capital_social_da_empresa')\
    .filter(f.upper(company_df['razao_social_nome_empresarial']).like('%RESTAURANTE%'))\
    .show(5, False)
```

```python
#summarize
bussiness_parter_df\
    .select(f.year('data_de_entrada_sociedade').alias('ano_de_entrada'))\
    .where('ano_de_entrada >= 2010')\
    .groupBy('ano_de_entrada')\
    .count()\
    .orderBy('ano_de_entrada', ascending=True)\
    .show()
```

```python
#agregate
company_df\
    .select('cnpj_basico', 'porte_da_empresa', 'capital_social_da_empresa')\
    .groupBy('porte_da_empresa')\
    .agg(
        f.avg('capital_social_da_empresa').alias('capital_social_medio'),
        f.count('cnpj_basico').alias('frequencia')
    )\
    .orderBy('porte_da_empresa', ascending=True)\
    .show()
```

```python
#summarize
company_df\
    .select('capital_social_da_empresa')\
    .summary()\
    .show()

# .summary('count', 'mean', 'stddev', 'min', '25%', '50%', '75%', 'max')
```

# Merge

```python
company_join_df = company_df.join(store_df, 'cnpj_basico', how='inner')
```

```python
company_join_df.printSchema()
```

```python
freq = company_join_df\
    .select(
            'cnpj_basico',
            f.year('data_de_inicio_atividade').alias('data_de_inicio')
    )\
    .where('data_de_inicio >= 2010')\
    .groupBy('data_de_inicio')\
    .agg(f.count('cnpj_basico').alias('freq'))\

freq.show()
```

```python
freq.union(
    freq.select(
        f.lit('total').alias('data_de_inicio'),
        f.sum(freq.freq).alias('freq')
    )
).show()
```

# SQL

```python
company_df.createOrReplaceTempView('company_view')
```

```python
spark.sql("SELECT * FROM company_view").show(5)
```

```python
spark\
    .sql("""
        SELECT * 
            FROM company_view
            WHERE capital_social_da_empresa = 50
        """)\
    .show(5)
```

```python
spark\
    .sql("""
        SELECT porte_da_empresa, MEAN(capital_social_da_empresa) AS media
            FROM company_view
            GROUP BY porte_da_empresa
        """)\
    .show(5)
```

```python
company_join_df.createOrReplaceTempView("company_join_view")
```

```python
freq = spark\
    .sql("""
        SELECT YEAR(data_de_inicio_atividade) AS data_de_inicio, COUNT(cnpj_basico) AS count
            FROM company_join_view 
            WHERE YEAR(data_de_inicio_atividade) >= 2010
            GROUP BY data_de_inicio
            ORDER BY data_de_inicio
    """)

freq\
    .show()
```

```python
freq = spark\
    .sql("""
        SELECT YEAR(data_de_inicio_atividade) AS data_de_inicio, COUNT(cnpj_basico) AS count
            FROM company_join_view 
            WHERE YEAR(data_de_inicio_atividade) >= 2010
            GROUP BY data_de_inicio
            ORDER BY data_de_inicio
    """)

freq\
    .show()
```

```python
freq.createOrReplaceTempView("freqView")
```

```python
spark\
    .sql("""
        SELECT *
            FROM freqView
        UNION ALL
        SELECT 'Total' AS data_de_inicio, SUM(count) AS count
            FROM freqView
    """)\
    .show()
```
